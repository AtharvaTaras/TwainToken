```python
import os, re
```

### Cleanup P1  





```python
filepath = os.path.join('data', 'original.txt')

with open(filepath, 'r', encoding='utf-8') as f:
    data = f.read()

print(data[1500:2000])
```

```python
pairs = [
    ['--', ' '], 
    ['__', ''], 
    ['“', '"'], 
    ['”', '"'],
]

data_p1 = data

for pair in pairs:
    data_p1 = re.sub(pair[0], pair[1], data_p1)

print(data_p1[1000:2000])
```

```python
p1_file = os.path.join('data', 'data_p1.txt')

with open(p1_file, 'w', encoding='utf-8') as f:
    f.write(data_p1)
```

### Cleanup P2
Punctuation marks   
Newlines  

```python
data_p2 = data_p1
data_p2 = re.sub(r'([!"#$%&\'()*+,\-./:;<=>?@[\\\]^_`{|}~])', r' \1 ', data_p2)
# data_p2 = re.sub(r'\n{2,}', ' <PARA>', data_p2)
# data_p2 = re.sub(r'\n', ' ', data_p2)
data_p2 = re.sub(r'\s+', ' ', data_p2)

print(data_p2[1000:2000])
```

```python
p2_file = os.path.join('data', 'data_p2.txt')

with open(p2_file, 'w', encoding='utf-8') as f:
    f.write(data_p2)
```
