```python
import os, re, json
```

```python
filepath = os.path.join('data', 'data_p3.txt')

with open(filepath, 'r', encoding='utf-8') as f:
    data = f.read()

print(data[1000:2000])
```

```python
tokens = data.split(' ')
unique = {}

for token in tokens:
    if token in unique:
        unique[token] += 1

    else:
        unique[token] = 1

unique[' '] = len(tokens)
```

```python
sorted_dict = dict(sorted(unique.items(), key=lambda x: x[1], reverse=True))
```

```python
sorted_dict
```

```python
len(sorted_dict)
```

```python
lookup = {}

for index, item in enumerate(sorted_dict.keys()):
    lookup[item] = index
```

```python
outfile = os.path.join('encodings', 'v2.json')

with open(outfile, 'w') as f:
    json.dump(lookup, f, indent=4)
```

```python
from random import randint

bank = 1000
start_bet = 100
min_bet = start_bet
place = 'even'
init = bank
exit = 1.5
win = False


def switch(place:str):
    if place == 'even':
        return 'odd'
    
    return 'even'


def spin():
    result = randint(0, 36)

    if result == 0:
        return 'zero'
    
    elif result % 2 == 0:
        return 'even'
    
    else:
        return 'odd'
    
   
while bank >= start_bet:
    if bank >= init * exit:
        print(f'Exited with {bank} after starting with {init}')
        win = True
        break

    bank -= start_bet
    result = spin()

    if result == place:
        bank += start_bet * 2
        place = switch(place)
        start_bet = min_bet

    else:
        start_bet *= 2

print(win)
```
